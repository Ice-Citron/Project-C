### Question 1: Redundant Helper Parameter

```c
uint64_t eval_op(uint32_t op, uint64_t a, uint64_t b, uint64_t mask) {
    switch (op) {
        case 0: 
            return (a + b) & mask; 
        case 1: 
            return (a - b) & mask; 
        default: 
            fprintf(stderr, "ERROR: Invalid opcode.\n");
            return UINT64_MAX;
    }
}
```
