```c
static int cmp_char(const void *a, const void *b) {
    char x = *(const char *)a;
    char y = *(const char *)b;
    return (x > y) - (x < y)
}

// Build signature
void make_signature(char *sig, const char *word) {
    strcpy(sig, word);
    qsort(sig, strlen(sig), sizeof(char), cmp_char);
}
```

```c
typedef struct region {
    int x, y, w, h, depth;
} region_t;

typedef struct node {
    region_t *region;
    struct node *prev;
    struct node *next;
} node_t

typedef struct list {
    node_t *head;
    node_t *tail;
} list_t;


// ---
list_t *list_create(void) {
    list_t *list = malloc(sizeof(*list));
    assert(list != NULL);

    list->head = malloc(sizeof(*list->head));
    list->tail = malloc(sizeof(*list->tail));
    assert(list->head != NULL);
    assert(list->tail != NULL);

    list->head->region = NULL;
    list->head->prev = NULL;
    list->head->next = list->tail;

    list->tail->region = NULL;
    list->tail->prev = list->head;
    list->tail->next = NULL;

    return list;
}
```

```c
// Insert before iterator
void list_insert_before(node_t *iter, region_t *region) {
    assert(iter != NULL);
    assert(region != NULL); 

    node_t *new = malloc( sizeof(*new) );
    assert(new != NULL);

    new->region = region;

    new->prev = iter->prev;
    new->next = iter;

    iter->prev->next = new;
    iter->prev = new;
}

void list_append(list_t *list, region_t *region) {
    list_insert_before(list->tail, region);
}

void list_delete_node(node_t *node) {
    assert(node != NULL);
    assert(node->prev != NULL);
    assert(node->next != NULL);

    node->prev->next = node->next;
    node->next->prev = node->prev;

    free(node->region);
    free(node);
}
```
