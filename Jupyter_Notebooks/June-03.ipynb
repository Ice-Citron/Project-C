   in C, the `static` keyword behaves like a split personality depending on
   exactly where you dropped it. When applied to a local variable inside a 
   function, it alters the variable's LIFETIME while preserving its narrow local
   scope. Instead of being dynamically allocated on the stack and wiped out 
   every time the function exits, a static local variable is placed in the
   permanent data segment of your program boots up and cleanly retains its 
   modified value across consecutive function calls. It acts like a secret 
   global variable that only that specific function is allowed to look at or
   change, which is incredibly useful for tracking state or tracking how many
   times an emulator loop has executed.

   When you move `static` outside of a function to a global variable or a 
   function declaration, its purpose completely flips: it restricts VISIBILITY
   (LINKAGE) instead of extending lifetime. A global static variable or function
   becomes strictly locked down and private to the individual `.c` file it 
   resides in. The rest of your repository cannot see it, link to it, or modify
   it. This serves as an architectural defense mechanism for an emulator
